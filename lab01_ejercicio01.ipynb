```python
# --- ESTADO INICIAL ---
# Solicitamos los valores al usuario y los convertimos a float
saldo = float(input("Ingresa el saldo inicial de la cuenta: "))
deposito = float(input("Ingresa el monto a depositar: "))
retiro = float(input("Ingresa el monto a retirar: "))

# --- ACTUALIZACIÓN DEL ESTADO ---
saldo = saldo + deposito  # Se suma el depósito al saldo
saldo = saldo - retiro    # Se resta el retiro al saldo

# 1. Declaración de nuevas variables para el interés
interes = 0.02  # 2% mensual
saldo_final = saldo + (saldo * interes) # Saldo actualizado con el interés aplicado

# --- SALIDA DEL ESTADO FINAL ---
# 2. Mostrar el resultado usando un f-string
print(f"Saldo final con interés: ${saldo_final:.2f}")
```

```python
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]
print("Lista inicial:", temperaturas)

temperaturas.append(26.2)
print("1. Agregar al final:", temperaturas)

temperaturas.pop(0)
print("2. Eliminar la primera:", temperaturas)

temperaturas.insert(1, 23.7)
print("3. Insertar 23.7 en Segunda posición:", temperaturas)

temperaturas.sort()
print("4. Lista ordenada:", temperaturas)

temp_max = max(temperaturas)
temp_min = min(temperaturas)
diferencia = temp_max - temp_min

promedio = sum(temperaturas) / len(temperaturas)
print(f"5. Diferencia max/min: {diferencia:.2f}°C | Promedio: {promedio:.2f}°C")

# Encuentra el índice (posición) de la temperatura 23.7.
indice_237 = temperaturas.index(23.7)
print(f"6. El índice de la temperatura 23.7 es: {indice_237}")
```

```python
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10},
]

print(" EJERCICIO 3 ")

print(f"1. El primer producto es: {inventario[0]['nombre']}")

# Agrega un nuevo producto al inventario
inventario.append({"nombre": "Galletas", "precio": 2000, "stock": 0})
print("2. Se ha agregado un nuevo producto (Galletas con stock 0).")

# Disminuye el stock de "Leche" en 2 unidades
for producto in inventario:
    if producto["nombre"] == "Leche":
        producto["stock"] -= 2
        break

# Calcula el valor total del inventario
valor_total = 0
for p in inventario:
    valor_total += p["precio"] * p["stock"]
print(f"4. El valor total del inventario es: ${valor_total}")

# 5. Muestra los productos cuyo stock sea menor a 10
productos_bajos = []
for p in inventario:
    if p["stock"] < 10:
        productos_bajos.append(p["nombre"])
print(f"5. Productos con stock menor a 10: {productos_bajos}")

# 6 y 7. Aumenta el precio en 10% y redondea
for p in inventario:
    # Se multiplica por 1.10 para subir 10% y se redondea a 2 decimales
    p["precio"] = round(p["precio"] * 1.10, 2)
print("6 y 7. Precios aumentados en 10% y redondeados.")

# 8. Búsqueda de un producto por nombre ingresado por el usuario
busqueda = input("Ingresa el nombre del producto que deseas buscar (ej. Pan): ")
# Usamos un bucle for para encontrar el primero que coincida (ignorando mayúsculas/minúsculas)
encontrado = None
for p in inventario:
    if p["nombre"].lower() == busqueda.lower():
        encontrado = p
        break

if encontrado:
    print(f"Producto encontrado: {encontrado}")
else:
    print("El producto no se encuentra en el inventario.")

# 9. Elimina un producto si su stock es 0 (Reescribimos la lista excluyendo los de stock 0)
new_inventario = []
for p in inventario:
    if p["stock"] > 0:
        new_inventario.append(p)
inventario = new_inventario
print("\n9. Se han eliminado los productos con stock 0 del sistema.")

# 10. Ordena la lista por precio ascendente
n = len(inventario)
for i in range(n):
    for j in range(0, n-i-1):
        if inventario[j]["precio"] > inventario[j+1]["precio"]:
            inventario[j], inventario[j+1] = inventario[j+1], inventario[j]

print(f"10. Inventario final ordenado por precio:")
for p in inventario:
    print(f"{p}")
```

```python
import math

print("--- INICIO EJERCICIO 4 ---")

punto = (3, 7)
print("Punto:", punto)

# 1. Crea una tupla rectangulo = (ancho, alto)
rectangulo = (5, 12)
ancho, alto = rectangulo

# 2. Calcula su área y perímetro
area = ancho * alto
perimetro = 2 * (ancho + alto)
print(f"Área del rectángulo: {area} | Perímetro: {perimetro}")

# 3. Intenta cambiar el valor de ancho
print("\nIntentando modificar el ancho del rectángulo...")
try:
    rectangulo[0] = 10
except TypeError as error_generado:
    print(f"--> ¡ERROR! Detalle: {error_generado}")

# ---------------------------------------------------------
posiciones = [(0, 0), (1, 2), (3, 4)]
distancia_total = 0

for i in range(1, len(posiciones)):
    x1, y1 = posiciones[i-1]
    x2, y2 = posiciones[i]

    distancia = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
    distancia_total += distancia

print(f"\nDistancia total recorrida entre los puntos: {distancia_total:.2f} unidades")
```

```python
print("--- INICIO EJERCICIO 5 ---")

biblioteca = {
    "libros": [
        {"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito", "autor": "Antoine de Saint-Exupéry", "prestado": True}
    ],
    "usuarios": ["Ana", "Luis", "María"]
}

# 1. Muestra cuántos libros están prestados y cuántos disponibles
prestados = 0
disponibles = 0
for libro in biblioteca["libros"]:
    if libro["prestado"] == True:
        prestados += 1
    else:
        disponibles += 1
print(f"1. Libros prestados: {prestados} | Libros disponibles: {disponibles}")

# 2. Agrega un nuevo libro
nuevo_libro = {"titulo": "Fahrenheit 451", "autor": "Ray Bradbury", "prestado": False}
biblioteca["libros"].append(nuevo_libro)
print("2. Se ha agregado un nuevo libro a la colección.")

# 3. Marca como prestado el libro "1984"
for libro in biblioteca["libros"]:
    if libro["titulo"] == "1984":
        libro["prestado"] = True
        print("3. El libro '1984' ha sido marcado como PRESTADO.")
        break

# 4. Muestra el estado actualizado del sistema
print("\n4. Estado final de la biblioteca:")
print(biblioteca)
```
